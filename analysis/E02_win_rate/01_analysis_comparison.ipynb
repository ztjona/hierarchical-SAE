# E02 Win Rate Experiments - Comparison Analysis

This notebook analyzes and compares the results from the E02_win_rate experiment variants with different learning rates:

- **base**: LR = 1e-4
- **A**: LR = 1e-5
- **B**: LR = 5e-5
- **C**: LR = 5e-4

```python
import sys
from pathlib import Path

sys.path.append(str(Path("../..").resolve()))
import os

if os.path.basename(os.getcwd()) == "tests":
    os.chdir("..")
```

```python
# Variant configurations
variant_configs = {
    "base": {"lr": 1e-4, "description": "Baseline"},
    "A": {"lr": 1e-5, "description": "Lower LR"},
    "B": {"lr": 5e-5, "description": "Medium-low LR"},
    "C": {"lr": 5e-4, "description": "Higher LR"},
}

print("Variant Configurations:")
for var, config in variant_configs.items():
    print(f"  {var}: LR = {config['lr']:.0e} ({config['description']})")
```

## Load Data from Pickle Files

```python
import pickle

variants = ["base", "A", "B", "C"]  # Include base variant
evals = {}

for variant in variants:
    # Use different filename for base variant
    file = f"E02_win_rate.pkl" if variant == "base" else f"E02_win_rate_{variant}.pkl"
    try:
        with open(file, "rb") as f:
            evals[variant] = pickle.load(f)
        print(f"Loaded {file}")
    except FileNotFoundError:
        print(f"Warning: {file} not found")

# Display structure of first variant
if evals:
    first_variant = list(evals.keys())[0]
    print(f"\nData structure for variant {first_variant}:")
    print(evals[first_variant].keys())
```

## Win Rate Comparison Across Variants

```python
import plotly.graph_objs as go

fig = go.Figure()

for variant, data in evals.items():
    win_rate = data["win_rate"]
    lr = variant_configs[variant]["lr"]
    for bot, rates in win_rate.items():
        fig.add_trace(
            go.Scatter(y=rates, mode="lines", name=f"{variant} (LR={lr:.0e}) - {bot}")
        )

fig.update_layout(
    title="Win Rate per Epoch - All Variants",
    xaxis_title="Epoch",
    yaxis_title="Win Rate",
    legend_title="Variant - Bot",
    height=600,
)
fig.show()
```

## Detailed Analysis per Variant (Win Rate + Loss)

```python
# Smooth win rates function
def smooth_win_rates(rates, window_size=11):
    smoothed = []
    for i in range(len(rates)):
        window = rates[
            max(0, i - window_size // 2) : min(len(rates), i + window_size // 2 + 1)
        ]
        smoothed.append(sum(window) / len(window))
    return smoothed


# Define consistent colors for each bot type
bot_colors = {
    "bot_good": "#1f77b4",  # Blue
    "bot_medium": "#ff7f0e",  # Orange
    "bot_random": "#2ca02c",  # Green
    "bot_bad": "#d62728",  # Red
}

# Create one subplot per variant
fig = make_subplots(
    rows=len(evals),
    cols=1,
    subplot_titles=[
        f"Variant {v} (LR={variant_configs[v]['lr']:.0e})" for v in evals.keys()
    ],
    vertical_spacing=0.08,
)

for i, (variant, data) in enumerate(evals.items(), 1):
    win_rate = data["win_rate"]
    lr = variant_configs[variant]["lr"]

    # Plot each bot's win rate
    for bot, rates in win_rate.items():
        bot_color = bot_colors.get(bot, "#999999")  # Default gray if bot not in dict

        # Raw data as markers
        fig.add_trace(
            go.Scatter(
                y=rates,
                mode="markers",
                name=f"{bot} (LR={lr:.0e}) - {variant}",
                marker=dict(size=3, color=bot_color, opacity=0.5),
                legendgroup=f"group{variant}",
                showlegend=True,
            ),
            row=i,
            col=1,
        )

        # Smoothed line
        smoothed = smooth_win_rates(rates)
        fig.add_trace(
            go.Scatter(
                y=smoothed,
                mode="lines",
                name=f"{bot} (LR={lr:.0e}) - {variant} (smooth)",
                line=dict(width=3, color=bot_color),
                # legendgroup=f"group{variant}",
                showlegend=False,
            ),
            row=i,
            col=1,
        )

    # Add loss
    loss_epoch = [
        data["loss_values"]["loss_values"][j]
        for j in data["loss_values"]["epoch_values"]
    ]
    fig.add_trace(
        go.Scatter(
            y=loss_epoch,
            mode="lines",
            name=f"{variant} (LR={lr:.0e})-Loss",
            line=dict(dash="dash"),
            legendgroup=f"group{variant}",
            showlegend=True,
        ),
        row=i,
        col=1,
    )
    fig.add_vline(x=20, line_width=1.5, line_dash="dash", line_color="red")

    # Update y-axis
    fig.update_yaxes(title_text="Win Rate / Loss", row=i, col=1)

# Update x-axis (only for bottom plot)
fig.update_xaxes(title_text="Epoch", row=len(evals), col=1)
fig.update_xaxes(range=[0, 400])

fig.update_layout(
    height=400 * len(evals),
    title_text="Detailed Win Rate and Loss Analysis per Variant",
)
fig.show()
```

```python
# Get all unique bots from the first variant
all_bots = list(evals[list(evals.keys())[0]]["win_rate"].keys())

# Create subplots - one per bot type
fig = make_subplots(
    rows=len(all_bots),
    cols=1,
    subplot_titles=[f"Win Rate vs {bot}" for bot in all_bots],
    vertical_spacing=0.1,
)

# Define colors for each variant
variant_colors = {
    "base": "#1f77b4",  # Blue
    "A": "#ff7f0e",  # Orange
    "B": "#2ca02c",  # Green
    "C": "#d62728",  # Red
}

for bot_idx, bot in enumerate(all_bots, 1):
    for variant, data in evals.items():
        win_rate = data["win_rate"]
        lr = variant_configs[variant]["lr"]

        if bot in win_rate:
            rates = win_rate[bot]
            variant_color = variant_colors.get(variant, "#999999")

            # Raw data as markers
            fig.add_trace(
                go.Scatter(
                    y=rates,
                    mode="markers",
                    name=f"{variant} (LR={lr:.0e})",
                    marker=dict(size=3, color=variant_color, opacity=0.3),
                    legendgroup=f"group{variant}",
                    showlegend=(bot_idx == 1),  # Only show legend for first subplot
                ),
                row=bot_idx,
                col=1,
            )

            # Smoothed line
            smoothed = smooth_win_rates(rates)
            fig.add_trace(
                go.Scatter(
                    y=smoothed,
                    mode="lines",
                    name=f"{variant} (LR={lr:.0e}) smooth",
                    line=dict(width=2, color=variant_color),
                    opacity=0.7,
                    legendgroup=f"group{variant}",
                    showlegend=False,
                ),
                row=bot_idx,
                col=1,
            )

    # Add vertical line at epoch 20
    fig.add_vline(
        x=27,
        line_width=1,
        line_dash="dash",
        line_color="gray",
        opacity=0.5,
        row=bot_idx,
        col=1,
    )

    # Update y-axis
    fig.update_yaxes(title_text="Win Rate", row=bot_idx, col=1)

# Update x-axis (only for bottom plot)
fig.update_xaxes(title_text="Epoch", row=len(all_bots), col=1)
# fig.update_xaxes(range=[1_000, 1300])
fig.update_xaxes(range=[0, 300])

fig.update_layout(
    height=300 * len(all_bots),
    title_text="Win Rate Comparison by Opponent Bot - All Variants",
)
fig.show()
```

## Statistical Summary

```python
import pandas as pd

# Prepare win rate data for correlation analysis
correlation_results = {}

for bot in win_rate.keys():
    # Collect win rates for each variant for the current bot
    win_rate_matrix = pd.DataFrame(
        {
            variant: evals[variant]["win_rate"][bot]
            for variant in variants
            if bot in evals[variant]["win_rate"]
        }
    )
    # Compute correlation matrix
    correlation_results[bot] = win_rate_matrix.corr()

# Display correlation matrices for each rival bot
for bot, corr_matrix in correlation_results.items():
    print(f"\nCorrelation matrix for {bot}:")
    print(corr_matrix.round(4))
```

```python
import pandas as pd
import numpy as np

stats = {}

for variant, data in evals.items():
    win_rate = data["win_rate"]
    stats[variant] = {}

    for bot, rates in win_rate.items():
        rates_array = np.array(rates)
        stats[variant][bot] = {
            "mean_win_rate": np.mean(rates_array),
            "final_win_rate": np.mean(rates_array[-20:]),  # Last 20 epochs
            "std_win_rate": np.std(rates_array),
            "max_win_rate": np.max(rates_array),
            "min_win_rate": np.min(rates_array),
            "total_epochs": len(rates_array),
        }

# Convert to DataFrame
stats_df = pd.DataFrame.from_dict(
    {
        (variant, bot): stats[variant][bot]
        for variant in stats.keys()
        for bot in stats[variant].keys()
    },
    orient="index",
)
stats_df.index.names = ["Variant", "Bot"]
stats_df = stats_df.round(4)
stats_df
```

## Training Log Analysis

```python
import re
from datetime import datetime

log_data = {}

for variant in variants:
    log_file = f"trainRL_e02_{variant}.log"
    try:
        with open(log_file, "r") as f:
            content = f.read()

        # Extract timestamps
        timestamps = re.findall(r"\[(.*?)\]", content)
        if timestamps:
            try:
                timestamps = [
                    datetime.strptime(ts, "%Y-%m-%d %H:%M:%S,%f") for ts in timestamps
                ]
                duration = (
                    timestamps[-1] - timestamps[0] if len(timestamps) > 1 else None
                )
            except:
                duration = None
        else:
            duration = None

        # Extract epoch count
        epochs = re.findall(r"Epoch (\d+)", content)
        total_epochs = len(set(epochs))

        log_data[variant] = {
            "total_epochs": total_epochs,
            "duration": duration,
            "log_lines": len(content.split("\n")),
        }

    except FileNotFoundError:
        print(f"Warning: {log_file} not found")
        log_data[variant] = {"total_epochs": 0, "duration": None, "log_lines": 0}

# Create summary DataFrame
log_summary = pd.DataFrame.from_dict(log_data, orient="index")
log_summary.index.name = "Variant"
log_summary
```
